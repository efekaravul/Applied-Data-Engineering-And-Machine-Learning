```python
from sklearn.model_selection import train_test_split
```

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 15)
```

```python
X_train.head()
```

```python
X_train = pd.get_dummies(X_train, columns=["brand"], drop_first=True, dtype=int)
X_test = pd.get_dummies(X_test, columns=["brand"], drop_first=True, dtype=int)
```

```python
X_train.head()
```

```python
model_price_map = y_train.groupby(X_train["model"]).mean()
X_train["model_target_enc"] = X_train["model"].map(model_price_map)
X_test["model_target_enc"] = X_test["model"].map(model_price_map)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[["Tier3", "Tier2", "Tier1"]])
X_train["city_tier_encoded"] = oe.fit_transform(X_train[["city_tier"]])

oe = OrdinalEncoder(categories=[["Tier3", "Tier2", "Tier1"]])
X_test["city_tier_encoded"] = oe.fit_transform(X_test[["city_tier"]])
```

```python
oe = OrdinalEncoder(categories=[["Poor","Fair","Good","Excellent"]])
X_train["condition_tier"] = oe.fit_transform(X_train[["condition"]])

oe = OrdinalEncoder(categories=[["Poor","Fair","Good","Excellent"]])
X_test["condition_tier"] = oe.fit_transform(X_test[["condition"]])
```

```python
X_train
```

```python
X_train = pd.get_dummies(X_train, columns=["os_type"], drop_first=True, dtype=int)
X_test = pd.get_dummies(X_test, columns=["os_type"], drop_first=True, dtype=int)
```

```python
X_train
```

```python
X_train = pd.get_dummies(X_train, columns=["seller_type"], drop_first=True, dtype=int)
X_test = pd.get_dummies(X_test, columns=["seller_type"], drop_first=True, dtype=int)
```

```python
X_train
```

```python
X_train.drop(columns=["model","condition","city_tier"], inplace=True)
```

```python
X_test.drop(columns=["model","condition","city_tier"], inplace=True)
```

```python
X_train
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression()
```

```python
regression.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
y_pred = regression.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, regression.predict(X_test))
mse = mean_squared_error(y_test, regression.predict(X_test))
r2 = r2_score(y_test, regression.predict(X_test))
```
