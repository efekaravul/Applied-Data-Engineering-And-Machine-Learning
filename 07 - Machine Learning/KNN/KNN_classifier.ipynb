```python
X = df.drop("high_risk_flag", axis=1)
y = df["high_risk_flag"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=15)
```

```python
sns.boxenplot(data=X_train)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
classifier = KNeighborsClassifier(n_neighbors=5, algorithm='auto', weights='uniform')
```

```python
classifier.fit(X_train_scaled, y_train)
```

```python
y_pred = classifier.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
classifier_2 = KNeighborsClassifier(n_neighbors=5, algorithm='kd_tree', weights='uniform')
classifier_2.fit(X_train_scaled, y_train)
```

```python
y_pred_2 = classifier_2.predict(X_test_scaled)
```

```python
print(confusion_matrix(y_test, y_pred_2))
print(classification_report(y_test, y_pred_2))
print(accuracy_score(y_test, y_pred_2))
```
