```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data1 = np.linspace(0,10,20)
data2 = data1 ** 2
```

```python
my_fig, my_axes = plt.subplots()
my_axes.plot(data1, data2, color="#8F301E", alpha=0.5, linestyle="--", linewidth=2)
my_axes.plot(data2, data1, color="#8F301E", alpha=0.5, linestyle="--", linewidth=2, marker="o", markersize=10, markerfacecolor="#8F301E")
plt.show()

```

```python
plt.scatter(data1, data2)
plt.show()
```
