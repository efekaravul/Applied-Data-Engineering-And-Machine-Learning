```python
import numpy as np
import matplotlib.pyplot as plt

age_list = [10,20,30,30,30,40,60,70,75]
weight_list = [20,60,80,85,86,87,70,90,95]

numpy_age = np.array(age_list)
numpy_weight = np.array(weight_list)

plt.plot(numpy_age, numpy_weight, "g")
plt.xlabel("Age")
plt.ylabel("Weight")
plt.title("Age vs Weight")
plt.show()
```

```python
numpy_arr = np.linspace(0,10,20)
numpy_arr_2 = numpy_arr ** 3

plt.plot(numpy_arr, numpy_arr_2, "r*-")
plt.show()
```
