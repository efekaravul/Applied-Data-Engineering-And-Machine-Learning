```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
numpy_arr1 = np.linspace(0,10,20)
numpy_arr2 = numpy_arr1 ** 3

plt.subplot(1,2,1) # 1 satır, 2 sütun, 1.grafik
plt.plot(numpy_arr1,numpy_arr2, "b*-")
plt.subplot(1,2,2) # 1 satır, 2 sütun, 2 grafik
plt.plot(numpy_arr2,numpy_arr1, "g--")
plt.show()
```

```python
my_figure = plt.figure()
figure_axes = my_figure.add_axes([0.1,0.1,0.7,0.7]) # 1- soldan 2- alttan 3- Genişlik 4- Yükseklik
figure_axes.plot(numpy_arr1,numpy_arr2, "b")
figure_axes.set_xlabel("X axis")
figure_axes.set_ylabel("Y axis")
figure_axes.set_title("Title")
plt.show()
```

```python
new_fig = plt.figure(dpi=100)
new_axes = new_fig.add_axes([0.1,0.1,0.9,0.9])
new_axes.plot(numpy_arr1,numpy_arr1**2, label=" ** 2")
new_axes.plot(numpy_arr1, numpy_arr1**4, label= "** 4")
new_axes.set_xlabel("X axis")
new_axes.set_ylabel("Y axis")
new_axes.set_title("Title")
new_axes.legend(loc=0)
plt.show()
```
