```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

Box Plot:
- Median
- Quartile
- Min Max
- Outlier
Değerlerini görebilmemizi sağlayan Seaborn styledır.

```python
data = np.array([1,2,5,7,9,10,25,20,30,54,32,56,90,70,82])

plt.figure(figsize=(6,5))
sns.boxplot(y=data)
plt.title('Box Plot Example')
plt.ylabel('Data Values')
plt.show()
```

```python
df = sns.load_dataset("titanic")
```

```python
df
```

```python
df.head()
df.tail()
```

```python
plt.figure(figsize=(15,8))

plt.subplot(1,2,1)
sns.boxplot(x="class", y="age", data=df)
plt.xlabel('Class')
plt.ylabel('Age')
plt.title("Age by Class")

plt.subplot(1,2,2)
sns.boxplot(x="class", y="fare", data=df)
plt.xlabel('Class')
plt.ylabel('Age')
plt.title("Fare by Class")

plt.tight_layout()
plt.show()
```

Kutu grafikleri (boxplot); verinin yığılımını, medyan (ortanca) değerini ve aykırı (istisnai) noktaları tek bakışta görmemizi sağlar. Renkli kutular verinin orta %50'lik kısmını, içlerindeki yatay çizgi ise tam orta değeri (medyan) temsil eder.

- Yaş ve Sınıf (Sol Grafik): 1. sınıf yolcuların yaş ortalaması (35-40 arası) diğer sınıflara göre belirgin şekilde daha yüksektir. 2. ve 3. sınıfa inildikçe yolcu kitlesi belirgin şekilde gençleşmektedir. Üstte kalan noktalar ise o sınıfın yaş ortalamasından çok daha büyük olan istisnai yolcuları gösterir.

- Ücret ve Sınıf (Sağ Grafik): 1. sınıf bilet fiyatlarında çok büyük dalgalanmalar vardır (500'ü aşan ekstrem bir bilet fiyatı bile mevcut). Buna karşılık 2. ve 3. sınıf bilet fiyatları tabana yayılmış, düşük ve birbirine çok yakın değerlerdedir.
